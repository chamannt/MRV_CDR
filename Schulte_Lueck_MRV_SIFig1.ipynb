### Data & cleaning

```python
REGEX_NON_ALPH = re.compile(r'[^a-z]')
def str_to_title_slug(title):
    if title is None or len(title.strip()) == 0:
        return None
    # remove all non-alphabetic characters
    return REGEX_NON_ALPH.sub('', title.lower())


df = pd.read_excel("MRVdata_1308.xlsx") 
df_filtered = df[df['Year'] != 2024]
df_included = df_filtered[df_filtered['Inclusion/exclusion'] == 'Inclusion'].copy()

### additonal data

def extract_CDR_method_focus(row):
    CDR_method = []
    CDR_focus = []
    
    for column in cdr_method_colors.keys():  # previously defined list with columns that contain CDR methods
        if row[column] > 0:
            CDR_method.append(column)
            CDR_focus.append(row[column])
    
    return [CDR_method, CDR_focus]
# add new columns to the data frame
df_included['CDR_focus'] = df_included.apply(lambda row: extract_CDR_method_focus(row), axis=1)
df_included['CDR_method'] = df_included.CDR_focus.str[0]
df_included['CDR_focus'] = df_included.CDR_focus.str[1]


#add new column for MRV topic and study focus to the dataset
def extract_MRV_topic_focus(row):
    MRV_topic = []
    MRV_focus = []
    
    for column in mrv_topic_colors.keys():  # previously defined list with columns that contain MRV topics
        if row[column] > 0:
            MRV_topic.append(column)
            MRV_focus.append(row[column])
    
    return [MRV_topic, MRV_focus]
# add new columns to the data frame
df_included['MRV_focus'] = df_included.apply(lambda row: extract_MRV_topic_focus(row), axis=1)
df_included['MRV_topic'] = df_included.MRV_focus.str[0]
df_included['MRV_focus'] = df_included.MRV_focus.str[1]


### explode
# explode CDR method and focus
df_CDRexplode = df_included.explode(['CDR_method', 'CDR_focus'])

# explode MRV topic and focus
df_MRVexplode = df_included.explode(['MRV_topic', 'MRV_focus'])
```

```python
# preparation of the different parts of the plot

df_CDRexploded = df.explode("CDR_method")
df_CDRexploded["CDR_method_norm"] = df_CDRexploded.CDR_method.apply(lambda x: x if x=='SCS' or x=="A/R" else 'other')

plot_df = df_CDRexploded.groupby(["Year", "is_included","CDR_method_norm"], as_index=False).size()
plot_df["keyword_exclusion"] = 0

ar_excluded_keywords = pd.read_excel("AR_relevant_sortedByYear.xlsx", sheet_name="excluded with keywords")
ar_excluded_keywords.rename(columns={"publication_year":"Year"}, inplace=True)
ar_ex_kw = ar_excluded_keywords.groupby("Year", as_index=False).size()
ar_ex_kw["CDR_method_norm"] = 'A/R'
ar_ex_kw["is_included"] = 0
ar_ex_kw["keyword_exclusion"] = 1

ar_included_keywords = pd.read_excel("AR_relevant_sortedByYear.xlsx", sheet_name="included with keywords")
ar_included_keywords.rename(columns={"publication_year":"Year"}, inplace=True)
ar_in_kw = ar_included_keywords.loc[ar_included_keywords["Year"]<2018].groupby("Year", as_index=False).size()
ar_in_kw["CDR_method_norm"] = 'A/R'
ar_in_kw["is_included"] = 0
ar_in_kw["keyword_exclusion"] = 0

scs_excluded_keywords = pd.read_excel("SCS_relevant_sortedByYear.xlsx", sheet_name="excluded with keywords")
scs_excluded_keywords.rename(columns={"publication_year":"Year"}, inplace=True)
scs_ex_kw = scs_excluded_keywords.groupby("Year", as_index=False).size()
scs_ex_kw["CDR_method_norm"] = 'SCS'
scs_ex_kw["is_included"] = 0
scs_ex_kw["keyword_exclusion"] = 1

scs_included_keywords = pd.read_excel("SCS_relevant_sortedByYear.xlsx", sheet_name="included with keywords")
scs_included_keywords.rename(columns={"publication_year":"Year"}, inplace=True)
scs_in_kw = scs_included_keywords.loc[scs_included_keywords["Year"]<2020].groupby("Year", as_index=False).size()
scs_in_kw["CDR_method_norm"] = 'SCS'
scs_in_kw["is_included"] = 0
scs_in_kw["keyword_exclusion"] = 0


plot_df = pd.concat([plot_df, ar_ex_kw, ar_in_kw, scs_ex_kw, scs_in_kw])

for y in plot_df.Year.unique():
    for incl in [0,1]:
        for cdr in plot_df.CDR_method_norm.unique():
            for keyword_exclusion in [0,1]:
                if len(plot_df[(plot_df.Year==y)& 
                               (plot_df.is_included==incl)&
                               (plot_df.CDR_method_norm==cdr)&
                               (plot_df.keyword_exclusion==keyword_exclusion)]) == 0:
                    one_row = pd.DataFrame({"Year":[y],
                                            "is_included":[incl],
                                            "CDR_method_norm":[cdr],
                                            "size":[0],
                                            "keyword_exclusion":[keyword_exclusion]})
                    plot_df = pd.concat([plot_df, one_row])
plot_df.sort_values(["Year","CDR_method_norm"], inplace=True)
```

### Panel B: AR Publication per years

```python
ar_col = { 0: '#238b45',
           1: '#74c476',
           2: '#bae4b3',
           3: '#edf8e9'}

ar_lab = {
    0:"checked and included",
    1:"checked and excluded",
    2:"excluded by keywords",
    3:"not checked and excluded"
}

def get_parameters(case, cutoff_year):
    if case==0:
        incl=1
        kw_ex=0
        plot_p = plot_df.copy()
        plot_p.loc[plot_p.Year<cutoff_year,"size"] = 0
    elif case==1:
        incl=0
        kw_ex=0
        plot_p = plot_df.copy()
        plot_p.loc[plot_p.Year<cutoff_year,"size"] = 0
    elif case==2:
        incl=0
        kw_ex=1
        plot_p = plot_df.copy()
    elif case==3:
        incl=0
        kw_ex=0
        plot_p = plot_df.copy()
        plot_p.loc[plot_p.Year>=cutoff_year,"size"] = 0
    return incl, kw_ex, plot_p

plot_df.drop_duplicates(subset=["Year","CDR_method_norm","is_included","keyword_exclusion"], inplace=True)

fig, ax = plt.subplots()

width=0.7
prev_values = np.zeros(len(plot_df.loc[(plot_df.is_included==0)&(plot_df.keyword_exclusion==0)&(plot_df.CDR_method_norm=="A/R"),"Year"]))
for case in range(4):
    incl, kw_ex, plot_p = get_parameters(case, 2018) 
        
    ax.bar( plot_p.loc[(plot_p.is_included==incl)&(plot_p.keyword_exclusion==kw_ex)&(plot_p.CDR_method_norm=="A/R"),"Year"],
            plot_p.loc[(plot_p.is_included==incl)&(plot_p.keyword_exclusion==kw_ex)&(plot_p.CDR_method_norm=="A/R"),"size"], 
            width, bottom=prev_values, 
            label=ar_lab[case],
            color=ar_col[case])
    prev_values += np.array(plot_p.loc[(plot_df.is_included==incl)&(plot_p.keyword_exclusion==kw_ex)&(plot_df.CDR_method_norm=="A/R"),"size"])

ax.legend()
```
